### Exploratory Wrangling


```python
conflict = df.loc[df['Conflict Present'] == 'Y', 'State'].str.split(';').explode()
conflict.unique()
```

### Find unique state codes

```python
state_conflict = conflict.str.strip().value_counts(dropna = False).reset_index()
state_conflict
```

#### Visualize data exploration & discussion

```python
state_conflict.plot(
   x='State',
   y='count',
   kind='bar',
   ylabel='# of Incidents',
   xlabel='States',
   color='pink',
   title='Number of Water Conflicts Per State' 
)
```
