Part B:  
i.) Bonzana Creek LTER scientitsts assembled this data. Snowshoe hares are a keystone prey in the norhtern boreal forests and experience serious population fluctuations and they want to understand it. 

ii.) A single row in the data represents one hare that was captured and tagged by the scientists.

iii.) These captures happened in 5 locations in Tanana valley from 1999 to 2002.

iv.) This data is not sensitive

v.) The experimental design of this study is described fully in Flora, B.K. 2002. Comparison of snowshoe hare populations in Interior Alaska. M.S. Thesis. University of Alaska-Fairbanks

vi.) The weight of the hare is recorded as a ratio.

vii.) It does not have any indication of how missing data is recorded. That row is left blank.

viii.) Yes this study was publishd in 2002. Here is the citation:
	
Flora, B.K. 2002. Comparison of snowshoe hare populations in Interior. M.S. Thesis. University of Alaska Fairbanks. Fairbanks, AK, USA.
